```python
import numpy as np
import matplotlib.pyplot as plt

def simulasi_pencerminan_sumbu_x():
    # 1. Definisikan Objek Asli (Segitiga seperti di gambar)
    # Titik-titik: D(-2, 1), Atas(1, 2), C(2, 1)
    objek = np.array([
        [-2, -1, 2, -2], # Koordinat X
        [ 1, 2, 1,  1]  # Koordinat Y
    ])

    # 2. Matriks Transformasi: Pencerminan Sumbu X
    # Rumus: x tetap, y berubah tanda (y -> -y)
    matriks_cermin_x = np.array([
        [1,  0],
        [0, -1]
    ])

    # 3. Hitung Hasil Transformasi (Perkalian Matriks)
    hasil_transformasi = matriks_cermin_x @ objek

    # 4. Visualisasi
    plt.figure(figsize=(10, 6))

    # Gambar Sumbu X dan Y sebagai referensi
    plt.axhline(0, color='black', linewidth=1.5, label='Garis Cermin (Sumbu X)')
    plt.axvline(0, color='black', linewidth=1)

    # Plot Objek Asli (Biru)
    plt.plot(objek[0], objek[1], 'bo-', label='Objek Asli (C)')
    plt.fill(objek[0], objek[1], 'blue', alpha=0.2)

    # Plot Hasil Pencerminan (Merah)
    plt.plot(hasil_transformasi[0], hasil_transformasi[1], 'ro-', label="Hasil Pencerminan (C')")
    plt.fill(hasil_transformasi[0], hasil_transformasi[1], 'red', alpha=0.2)

    # Pengaturan Tampilan Grafik
    plt.title("Visualisasi Transformasi Matriks: Pencerminan Sumbu X")
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.legend()
    plt.axis('equal') # Memastikan skala x dan y sama agar tidak distorsi
    plt.xlim(-4, 4)
    plt.ylim(-3, 3)

    # Tampilkan koordinat pada titik-titik (opsional)
    for i in range(3):
        plt.text(objek[0,i], objek[1,i], f'({objek[0,i]}, {objek[1,i]})', va='bottom')
        plt.text(hasil_transformasi[0,i], hasil_transformasi[1,i],
                 f'({hasil_transformasi[0,i]}, {hasil_transformasi[1,i]})', va='top')

    plt.show()

# Jalankan fungsi
simulasi_pencerminan_sumbu_x()
```
