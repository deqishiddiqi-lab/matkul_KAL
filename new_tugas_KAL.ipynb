```python
import numpy as np

def qr_iteration(A, iterations=10):
    """
    Melakukan iterasi QR pada matriks A sebanyak n kali.
    """
    Ak = A.copy()

    print(f"Matriks Awal:\n{Ak}\n")
    print("-" * 30)

    for i in range(1, iterations + 1):
        # Langkah 1: Dekomposisi QR
        Q, R = np.linalg.qr(Ak)

        # Langkah 2: Hitung matriks berikutnya (Ak+1 = R * Q)
        Ak = np.dot(R, Q)

        print(f"Iterasi ke-{i}:")
        print(Ak)
        print("-" * 30)

    return Ak

# Contoh penggunaan:
# Mendefinisikan matriks persegi (n x n)
A = np.array([[4, 1, -1],
              [1, 2, 1],
              [-1, 1, 3]], dtype=float)

# Menjalankan fungsi
final_matrix = qr_iteration(A, 10)

print("\nHasil Akhir (Nilai Eigen mendekati elemen diagonal):")
print(np.diag(final_matrix))
```
